```python
import pandas as pd

series = pd.Series([80, 90, 75])
df = pd.DataFrame({
    "Name": ["Ram", "Sam", "Ravi"],
    "Marks": [80, 90, 75]
})

print(series)
print(df)
```

```python
df1 = pd.DataFrame({
    "ID": [1, 2],
    "Name": ["Ram", "Sam"]
})

df2 = pd.DataFrame({
    "ID": [1, 2],
    "Marks": [85, 90]
})

result = pd.merge(df1, df2, on="ID")
print(result)
```

```python
df = pd.DataFrame({
    "Department": ["IT", "HR", "IT", "HR"],
    "Marks": [80, 70, 90, 75]
})

print(df.groupby("Department")["Marks"].mean())
```

```python
df = pd.DataFrame({
    "Name": ["Ram", "Sam", "Ravi"],
    "Marks": [80, 70, 90],
    "Age": [21, 22, 19]
})

result = df[(df["Marks"] > 75) & (df["Age"] > 20)]
print(result)
```

```python
df = pd.DataFrame({
    "Name": ["Ram", "Sam", "Ravi"],
    "Marks": [80, np.nan, 90]
})

print(df.isnull().sum())
df["Marks"] = df["Marks"].fillna(df["Marks"].mean())
print(df)
```

```python
arr = np.array([10, 20, 30, 40, 50])

print("Sum:", np.sum(arr))
print("Mean:", np.mean(arr))
print("Median:", np.median(arr))
print("Variance:", np.var(arr))
print("Standard Deviation:", np.std(arr))
```

```python
df = pd.DataFrame({
    "Age": [20, 25, 30, 35, 40],
    "Department": ["IT", "HR", "IT", "HR", "IT"]
})

print("Before optimization:")
print(df.memory_usage(deep=True).sum())

df["Age"] = df["Age"].astype("int8")
df["Department"] = df["Department"].astype("category")

print("After optimization:")
print(df.memory_usage(deep=True).sum())
```
