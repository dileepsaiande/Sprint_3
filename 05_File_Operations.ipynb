# Reading and Writing Data

Pandas provides functions for reading and writing data in different file formats.

In this notebook, we will learn how to work with CSV, Excel, JSON, Parquet, Pickle, and SQL data.

We will also learn how to read multiple files and handle file paths using Python.

```python
import pandas as pd
import numpy as np
from pathlib import Path
```

```python
data = {
    "Name": ["Alice", "Bob", "Charlie", "David"],
    "Age": [20, 21, 22, 23],
    "Marks": [85, 90, 78, 92]
}

df = pd.DataFrame(data)

print(df)
```

## 1. CSV Files

CSV stands for Comma-Separated Values.

CSV files are commonly used to store tabular data.

Pandas provides `read_csv()` to read CSV files and `to_csv()` to write DataFrames to CSV files.

```python
df.to_csv("students.csv", index=False)

print("CSV file created successfully.")
```

```python
csv_data = pd.read_csv("students.csv")

print(csv_data)
```

```python
print("Shape:", csv_data.shape)
print("Columns:", csv_data.columns.tolist())
```

## 2. Excel Files

Excel files are commonly used for storing structured tabular data.

Pandas provides `read_excel()` for reading Excel files and `to_excel()` for writing DataFrames to Excel files.

```python
df.to_excel("students.xlsx", index=False)

print("Excel file created successfully.")
```

```python
excel_data = pd.read_excel("students.xlsx")

print(excel_data)
```

## 3. JSON Files

JSON stands for JavaScript Object Notation.

JSON is commonly used for storing and exchanging structured data.

Pandas provides `read_json()` and `to_json()` for working with JSON files.

```python
df.to_json("students.json", orient="records", indent=4)

print("JSON file created successfully.")
```

```python
json_data = pd.read_json("students.json")

print(json_data)
```

## 4. Parquet Files

Parquet is a column-oriented file format commonly used for large datasets and data engineering.

It provides efficient storage and fast reading of structured data.

```python
df.to_parquet("students.parquet", index=False)

print("Parquet file created successfully.")
```

```python
parquet_data = pd.read_parquet("students.parquet")

print(parquet_data)
```

## 5. Pickle Files

Pickle is a Python serialization format.

It can store Python objects and restore them later.

Pandas provides `to_pickle()` and `read_pickle()` for working with Pickle files.
