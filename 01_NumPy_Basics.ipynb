# Introduction to NumPy

NumPy stands for Numerical Python. It is a Python library used for numerical computing.

NumPy provides powerful data structures and functions for working with arrays, mathematical operations, statistics, linear algebra, and scientific computing.

In this notebook, we will learn the basic concepts of NumPy arrays and their operations.

```python
import numpy as np
```

## 1. What is NumPy?

NumPy is an open-source Python library mainly used for numerical and scientific computing.

It provides a powerful array object called `ndarray`, which allows us to store and process large amounts of numerical data efficiently.

```python
import numpy as np

arr = np.array([10, 20, 30, 40, 50])

print(arr)
print(type(arr))
```

## 2. Why NumPy?

NumPy is useful because:

- It provides fast numerical operations.
- It supports multidimensional arrays.
- It uses less memory than traditional Python lists for numerical data.
- It provides many mathematical functions.
- It is widely used in Data Science, Machine Learning, AI, and scientific computing.

```python
python_list = [1, 2, 3, 4, 5]

numpy_array = np.array([1, 2, 3, 4, 5])

print("Python List:", python_list)
print("NumPy Array:", numpy_array)
```

## 3. NumPy Arrays (ndarray)

A NumPy array is called an `ndarray`, which stands for N-dimensional array.

It can store data in one or more dimensions.

```python
arr = np.array([10, 20, 30, 40])

print(arr)
print("Type:", type(arr))
print("Data type:", arr.dtype)
```

## 4. Creating Arrays

NumPy provides several ways to create arrays, such as:

- `np.array()`
- `np.zeros()`
- `np.ones()`
- `np.arange()`
- `np.linspace()`

```python
# Using np.array()
arr1 = np.array([1, 2, 3, 4, 5])

# Using np.zeros()
arr2 = np.zeros(5)

# Using np.ones()
arr3 = np.ones(5)

# Using np.arange()
arr4 = np.arange(1, 10)

# Using np.linspace()
arr5 = np.linspace(1, 10, 5)

print("array:", arr1)
print("zeros:", arr2)
print("ones:", arr3)
print("arange:", arr4)
print("linspace:", arr5)
```

## 5. Dimensions

The dimension of a NumPy array represents the number of levels in the array.

Examples:

- 0-D array → single value
- 1-D array → list of values
- 2-D array → rows and columns
- 3-D array → collection of 2-D arrays

```python
arr_0d = np.array(10)

arr_1d = np.array([1, 2, 3, 4])

arr_2d = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

print("0-D:", arr_0d.ndim)
print("1-D:", arr_1d.ndim)
print("2-D:", arr_2d.ndim)
```

## 6. Shape

The `shape` attribute returns the size of an array in each dimension.

```python
arr = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

print("Array:")
print(arr)

print("Shape:", arr.shape)
```

## 7. Size

The `size` attribute returns the total number of elements in an array.

```python
arr = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

print("Total elements:", arr.size)
```

## 8. Data Types

NumPy arrays have a data type that specifies the type of values stored in the array.

Common NumPy data types include:

- int
- float
- bool
- str

```python
int_array = np.array([1, 2, 3])
float_array = np.array([1.1, 2.2, 3.3])
bool_array = np.array([True, False, True])

print("Integer:", int_array.dtype)
print("Float:", float_array.dtype)
print("Boolean:", bool_array.dtype)
```

## 9. Indexing

Indexing is used to access individual elements of an array.

NumPy uses zero-based indexing, meaning the first element has index 0.

```python
arr = np.array([10, 20, 30, 40, 50])

print("First element:", arr[0])
print("Third element:", arr[2])
print("Last element:", arr[-1])
```

## 10. Slicing

Slicing is used to access a range of elements from an array.

The syntax is:

`array[start:stop:step]`

```python
arr = np.array([10, 20, 30, 40, 50])

print("First three:", arr[0:3])
print("From second element:", arr[1:])
print("Every second element:", arr[::2])
```

## 11. Reshaping

Reshaping changes the structure of an array without changing its data.

```python
arr = np.arange(1, 13)

print("Original:")
print(arr)

reshaped = arr.reshape(3, 4)

print("Reshaped:")
print(reshaped)
```

## 12. Flatten

The `flatten()` method converts a multidimensional array into a one-dimensional array.

```python
arr = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

flat = arr.flatten()

print("Original:")
print(arr)

print("Flattened:")
print(flat)
```

## 13. Copy vs View

A copy creates a completely separate array.

A view creates another way of looking at the same underlying data.

Changes made to a copy do not affect the original array, while changes made through a view can affect the original array.

```python
# Copy
arr = np.array([1, 2, 3, 4])

copy_arr = arr.copy()
copy_arr[0] = 100

print("Original after copy modification:", arr)
print("Copy:", copy_arr)
```

```python
# View
arr = np.array([1, 2, 3, 4])

view_arr = arr.view()
view_arr[0] = 100

print("Original after view modification:", arr)
print("View:", view_arr)
```

## 14. Broadcasting

Broadcasting allows NumPy to perform operations between arrays of different shapes when their dimensions are compatible.

```python
arr = np.array([10, 20, 30, 40])

result = arr + 5

print("Original:", arr)
print("After adding 5:", result)
```

## 15. Array Operations

NumPy supports arithmetic operations directly on arrays.

Common operations include:

- Addition
- Subtraction
- Multiplication
- Division
- Power

```python
a = np.array([10, 20, 30])
b = np.array([2, 4, 5])

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Power:", a ** 2)
```

## 16. Universal Functions (ufuncs)

Universal functions, or ufuncs, are NumPy functions that operate element-by-element on arrays.

Examples include:

- `np.add()`
- `np.subtract()`
- `np.multiply()`
- `np.sqrt()`
- `np.exp()`
- `np.sin()`

```python
arr = np.array([1, 4, 9, 16])

print("Square root:", np.sqrt(arr))
print("Exponential:", np.exp(arr))
```

```python
a = np.array([10, 20, 30])
b = np.array([1, 2, 3])

print("Addition:", np.add(a, b))
print("Subtraction:", np.subtract(a, b))
print("Multiplication:", np.multiply(a, b))
```
