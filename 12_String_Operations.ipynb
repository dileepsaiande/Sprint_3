# String Operations with Pandas

String operations are commonly used for cleaning and transforming text data.

Pandas provides string methods through the `.str` accessor.

In this notebook, we will learn how to convert strings to lowercase and uppercase, remove spaces, split strings, search for text, replace values, and use regular expressions.

```python
import pandas as pd
import re
```

```python
data = {
    "Name": [
        " Alice ",
        "BOB",
        " Charlie ",
        "David",
        "EVA"
    ],
    "City": [
        "Hyderabad",
        "Chennai",
        "Bangalore",
        "Hyderabad",
        "Chennai"
    ],
    "Email": [
        "alice@gmail.com",
        "bob@yahoo.com",
        "charlie@gmail.com",
        "david@outlook.com",
        "eva@gmail.com"
    ]
}

df = pd.DataFrame(data)

print(df)
```

## 1. lower()

The `lower()` method converts all characters in a string to lowercase.

In Pandas, we use `.str.lower()` to apply it to a column.

It is useful for standardizing text before comparison or analysis.

```python
df["Name_Lower"] = df["Name"].str.lower()

print(df[["Name", "Name_Lower"]])
```

## 2. upper()

The `upper()` method converts all characters in a string to uppercase.

In Pandas, we use `.str.upper()` for a Series or column.

It is useful for standardizing text formats.

```python
df["Name_Upper"] = df["Name"].str.upper()

print(df[["Name", "Name_Upper"]])
```

## 3. strip()

The `strip()` method removes leading and trailing whitespace from strings.

In Pandas, we use `.str.strip()`.

It is commonly used during data cleaning because unwanted spaces can cause incorrect comparisons.

```python
df["Clean_Name"] = df["Name"].str.strip()

print(df[["Name", "Clean_Name"]])
```

```python
df["Clean_Name"] = (
    df["Name"]
    .str.strip()
    .str.lower()
)

print(df[["Name", "Clean_Name"]])
```

## 4. split()

The `split()` method divides a string into multiple parts based on a separator.

In Pandas, we use `.str.split()`.

It is useful for separating names, addresses, words, or other structured text.

```python
names = pd.Series([
    "Alice Smith",
    "Bob Kumar",
    "Charlie Jones"
])

split_names = names.str.split()

print(split_names)
```

```python
name_parts = names.str.split(" ", expand=True)

print(name_parts)
```

## 5. contains()

The `contains()` method checks whether a string contains a specified pattern or text.

In Pandas, we use `.str.contains()`.

It returns True or False for each value.

```python
gmail_users = df[df["Email"].str.contains("gmail")]

print(gmail_users)
```

```python
result = df[
    df["City"].str.contains("HYDERABAD", case=False)
]

print(result)
```

## 6. replace()

The `replace()` method is used to replace specific text with another value.

In Pandas, `.str.replace()` can modify text inside strings.

It is useful for correcting or standardizing text data.

```python
df["City_Clean"] = df["City"].str.replace(
    "Hyderabad",
    "HYD"
)

print(df[["City", "City_Clean"]])
```

```python
df["Email_Clean"] = df["Email"].str.replace(
    "@gmail.com",
    "@example.com",
    regex=False
)

print(df[["Email", "Email_Clean"]])
```

## 7. Regular Expressions (Regex)

Regular expressions are patterns used to search, extract, validate, or replace text.

Pandas string methods can use regular expressions for advanced text processing.

Regex is especially useful for emails, phone numbers, IDs, and other structured text.

```python
numbers = pd.Series([
    "Order 123",
    "Order 456",
    "Order 789"
])

result = numbers.str.extract(
    r"(\d+)"
)

print(result)
```

```python
emails = pd.Series([
    "alice@gmail.com",
    "bob@yahoo.com",
    "charlie@gmail.com"
])

domains = emails.str.extract(
    r"@(.+)$"
)

print(domains)
```

```python
email_pattern = r"^[\w\.-]+@[\w\.-]+\.\w+$"

valid_emails = emails.str.match(email_pattern)

print(valid_emails)
```

## Practical Example

Let's clean a customer dataset using multiple string operations.

We will remove extra spaces, convert names to lowercase, standardize cities, and identify Gmail users.

```python
customers = pd.DataFrame({
    "Name": [
        " Alice ",
        "BOB ",
        " Charlie",
        "David "
    ],
    "City": [
        "Hyderabad",
        "hyderabad",
        "Chennai",
        "Bangalore"
    ],
    "Email": [
        "alice@gmail.com",
        "bob@yahoo.com",
        "charlie@gmail.com",
        "david@outlook.com"
    ]
})

customers["Name"] = (
    customers["Name"]
    .str.strip()
    .str.lower()
)

customers["City"] = (
    customers["City"]
    .str.strip()
    .str.title()
)

customers["Is_Gmail"] = (
    customers["Email"]
    .str.contains("@gmail.com", regex=False)
)

print(customers)
```

## Combining String Operations

Multiple string operations can be chained together to clean and standardize text data.

```python
text = pd.Series([
    "  HELLO WORLD  ",
    "  Python Pandas  ",
    "  DATA SCIENCE  "
])

clean_text = (
    text
    .str.strip()
    .str.lower()
    .str.replace(" ", "_", regex=False)
)

print(clean_text)
```

## Common Pandas String Methods

| Method | Purpose |
|---|---|
| `str.lower()` | Converts text to lowercase |
| `str.upper()` | Converts text to uppercase |
| `str.strip()` | Removes leading/trailing spaces |
| `str.split()` | Splits strings |
| `str.contains()` | Searches for text |
| `str.replace()` | Replaces text |
| `str.extract()` | Extracts patterns using Regex |
| `str.match()` | Checks whether text matches a Regex pattern |
