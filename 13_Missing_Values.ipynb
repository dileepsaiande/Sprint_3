# Notebook 13 – Missing Values

## Introduction
Missing values are data entries that are empty or unavailable.
Pandas represents missing values using NaN, None, or pd.NA.
Handling missing values is important for accurate data analysis.
In this notebook, we will learn different missing value handling techniques.

```python
import pandas as pd
import numpy as np
```

## Creating a Dataset with Missing Values

We create a sample dataset containing missing values.
This dataset will help us practice different handling techniques.

```python
data = {
    "Name": ["Alice", "Bob", "Charlie", "David", "Eva"],
    "Age": [25, np.nan, 30,  None, 35],
    "Salary": [40000, 50000, np.nan, 60000, None],
    "City": ["Hyderabad", None, "Chennai", "Mumbai", "Hyderabad"]
}

df = pd.DataFrame(data)

print(df)
```

## 1. isnull()

The isnull() method identifies missing values.
It returns True for missing values and False for available values.
It helps us identify missing data in a DataFrame.

```python
print(df.isnull())
```

```python
print(df.isnull().sum())
```

## 2. notnull()

The notnull() method identifies values that are not missing.
It returns True for available values and False for missing values.
It is useful for selecting records with valid data.

```python
print(df.notnull())
```

```python
print(df[df["Age"].notnull()])
```

## 3. dropna()

The dropna() method removes rows or columns containing missing values.
By default, it removes rows containing at least one missing value.
It is useful when incomplete records should be excluded.

```python
print(df.dropna())
```

```python
print(df.dropna(how="all"))
```

## 4. fillna()

The fillna() method replaces missing values with specified values.
We can use a constant, mean, median, or mode.
The replacement strategy depends on the type of data.

```python
df_filled = df.copy()

df_filled["Age"] = df_filled["Age"].fillna(0)
df_filled["Salary"] = df_filled["Salary"].fillna(0)
df_filled["City"] = df_filled["City"].fillna("Unknown")

print(df_filled)
```

```python
df_stats = df.copy()

df_stats["Age"] = df_stats["Age"].fillna(
    df_stats["Age"].mean()
)

df_stats["Salary"] = df_stats["Salary"].fillna(
    df_stats["Salary"].median()
)

print(df_stats)
```

## 5. interpolate()

The interpolate() method estimates missing numerical values from surrounding values.
It is useful for continuous data and time-series datasets.
Linear interpolation estimates values along a straight line between known points.

```python
numbers = pd.Series([10, 20, np.nan, 40, 50])

result = numbers.interpolate()

print("Original values:")
print(numbers)

print("After interpolation:")
print(result)
```

## 6. Missing Value Handling Strategies

Different strategies are used depending on the data and the reason values are missing.

1. Remove rows or columns using dropna().
2. Replace missing values using fillna().
3. Use mean, median, or mode for suitable columns.
4. Use interpolation for continuous or time-series data.
5. Investigate why data is missing before choosing a strategy.

```python
# Create a copy for cleaning
cleaned_df = df.copy()

# Fill numerical missing values with their median
cleaned_df["Age"] = cleaned_df["Age"].fillna(
    cleaned_df["Age"].median()
)

cleaned_df["Salary"] = cleaned_df["Salary"].fillna(
    cleaned_df["Salary"].median()
)

# Fill missing categorical values with the mode
cleaned_df["City"] = cleaned_df["City"].fillna(
    cleaned_df["City"].mode()[0]
)

print(cleaned_df)
print("\nRemaining missing values:")
print(cleaned_df.isnull().sum())
```

## Practical Example – Customer Data Cleaning

We will inspect missing values in a customer dataset and handle them.
Numerical columns will use median imputation.
Categorical columns will use the most frequent value.

```python
customers = pd.DataFrame({
    "Customer": ["A", "B", "C", "D", "E"],
    "Age": [22, 28, np.nan, 35, 30],
    "Purchase": [1000, np.nan, 1500, 2000, 1800],
    "City": ["Hyderabad", "Chennai", None, "Chennai", "Mumbai"]
})

print("Before cleaning:")
print(customers)

customers["Age"] = customers["Age"].fillna(
    customers["Age"].median()
)

customers["Purchase"] = customers["Purchase"].fillna(
    customers["Purchase"].median()
)

customers["City"] = customers["City"].fillna(
    customers["City"].mode()[0]
)

print("\nAfter cleaning:")
print(customers)

print("\nMissing value counts:")
print(customers.isnull().sum())
```
