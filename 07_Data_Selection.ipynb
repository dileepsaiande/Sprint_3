# Selecting and Filtering Data

Pandas provides several methods for selecting and filtering data from a DataFrame.

In this notebook, we will learn how to select rows and columns using `loc` and `iloc`, filter data using Boolean conditions, and use functions such as `query()`, `isin()`, and `between()`.

We will also learn how to sort data using Pandas.

```python
import pandas as pd
```

```python
data = {
    "Name": ["Alice", "Bob", "Charlie", "David", "Eva", "Frank"],
    "Age": [20, 21, 22, 23, 24, 25],
    "City": ["Hyderabad", "Chennai", "Hyderabad", "Bangalore", "Chennai", "Hyderabad"],
    "Marks": [85, 90, 78, 92, 88, 76]
}

df = pd.DataFrame(data)

print(df)
```

## 1. loc

The `loc` method is used to select rows and columns using labels.

It can select specific rows, columns, or both based on their labels.

```python
print(df.loc[0])
```

```python
print(df.loc[0:2])
```

```python
print(df.loc[:, ["Name", "Marks"]])
```

```python
print(df.loc[0:2, ["Name", "Marks"]])
```

## 2. iloc

The `iloc` method is used to select rows and columns based on their integer positions.

It uses zero-based indexing.

```python
print(df.iloc[0])
```

```python
print(df.iloc[0:3])
```

```python
print(df.iloc[:, 0:2])
```

```python
print(df.iloc[0:3, 0:2])
```

## 3. Boolean Indexing

Boolean indexing filters data using conditions that return True or False.

It is commonly used to select rows that satisfy a specific condition.

```python
high_marks = df[df["Marks"] > 85]

print(high_marks)
```

```python
result = df[(df["Age"] > 20) & (df["Marks"] > 80)]

print(result)
```

## 4. Filtering Rows

Filtering rows means selecting only the records that satisfy a particular condition.

```python
hyderabad_students = df[df["City"] == "Hyderabad"]

print(hyderabad_students)
```

```python
passed_students = df[df["Marks"] >= 80]

print(passed_students)
```

```python
young_students = df[df["Age"] <= 22]

print(young_students)
```

## 5. query()

The `query()` method provides a convenient way to filter rows using a string expression.

It can make filtering conditions easier to read.

```python
result = df.query("Marks > 85")

print(result)
```

```python
result = df.query("Age > 20 and Marks > 80")

print(result)
```

```python
result = df.query("City == 'Hyderabad'")

print(result)
```

## 6. isin()

The `isin()` method checks whether values belong to a specified list of values.

It is useful when we want to filter rows based on multiple possible values.

```python
cities = ["Hyderabad", "Chennai"]

result = df[df["City"].isin(cities)]

print(result)
```

```python
names = ["Alice", "David"]

result = df[df["Name"].isin(names)]

print(result)
```

## 7. between()

The `between()` method selects values that fall within a specified range.

By default, the boundaries are included.

```python
result = df[df["Age"].between(21, 24)]

print(result)
```

```python
result = df[df["Marks"].between(80, 90)]

print(result)
```

## 8. Sorting Data

Pandas provides the `sort_values()` method to sort rows based on one or more columns.

We can sort data in ascending or descending order.

```python
sorted_df = df.sort_values("Marks")

print(sorted_df)
```

```python
sorted_df = df.sort_values("Marks", ascending=False)

print(sorted_df)
```

```python
sorted_df = df.sort_values("Age")

print(sorted_df)
```

```python
sorted_df = df.sort_values(
    ["City", "Marks"],
    ascending=[True, False]
)

print(sorted_df)
```

## Practical Example

Let's combine selection and filtering techniques to find students based on their city, age, and marks.

```python
result = df[
    (df["City"] == "Hyderabad") &
    (df["Marks"] >= 80)
]

print("Hyderabad students with marks >= 80:")
print(result)
```

```python
result = df.query(
    "Age >= 21 and Marks >= 80"
)

print(result)
```
