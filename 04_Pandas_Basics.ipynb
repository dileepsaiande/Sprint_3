# Introduction to Pandas

Pandas is a powerful Python library used for data manipulation and data analysis.

It provides easy-to-use data structures such as Series and DataFrame.

In this notebook, we will learn the basics of Pandas, including Series, DataFrames, data types, indexing, columns, reading data, and writing data.

```python
import pandas as pd
```

## 1. What is Pandas?

Pandas is an open-source Python library used for data manipulation and analysis.

It provides powerful data structures such as Series and DataFrame for handling structured data.

Pandas is commonly used in Data Science, Machine Learning, data cleaning, and data analysis.

```python
import pandas as pd

print("Pandas version:", pd.__version__)
```

## 2. Why Pandas?

Pandas is useful for:

- Reading and writing datasets
- Cleaning data
- Filtering and selecting data
- Handling missing values
- Analyzing structured data
- Working with CSV, Excel, JSON, and other file formats

```python
data = [10, 20, 30, 40, 50]

series = pd.Series(data)

print(series)
```

## 3. Series

A Series is a one-dimensional labeled data structure in Pandas.

It can contain numbers, strings, or other Python objects.

Each value in a Series has an index.

```python
marks = pd.Series([85, 90, 78, 92, 88])

print(marks)
```

```python
marks = pd.Series(
    [85, 90, 78],
    index=["Student1", "Student2", "Student3"]
)

print(marks)
```

## 4. DataFrame

A DataFrame is a two-dimensional labeled data structure in Pandas.

It consists of rows and columns and is similar to a table in a database or spreadsheet.

```python
data = {
    "Name": ["Alice", "Bob", "Charlie"],
    "Age": [20, 21, 22],
    "Marks": [85, 90, 88]
}

df = pd.DataFrame(data)

print(df)
```

## 5. Creating DataFrames

A DataFrame can be created from dictionaries, lists, NumPy arrays, and other data structures.

```python
data = {
    "Name": ["John", "Sara", "Mike"],
    "Age": [25, 24, 26],
    "City": ["Hyderabad", "Chennai", "Bangalore"]
}

df = pd.DataFrame(data)

print(df)
```

```python
data = [
    {"Name": "John", "Age": 25},
    {"Name": "Sara", "Age": 24},
    {"Name": "Mike", "Age": 26}
]

df = pd.DataFrame(data)

print(df)
```

## 6. Data Types

Pandas automatically assigns data types to columns based on the values they contain.

The `dtypes` attribute can be used to view the data types of DataFrame columns.

```python
data = {
    "Name": ["Alice", "Bob", "Charlie"],
    "Age": [20, 21, 22],
    "Marks": [85.5, 90.0, 88.5]
}

df = pd.DataFrame(data)

print(df.dtypes)
```

## 7. Index

The index identifies each row in a Pandas Series or DataFrame.

By default, Pandas uses numeric indexes starting from 0.

We can also create custom indexes.

```python
data = {
    "Name": ["Alice", "Bob", "Charlie"],
    "Marks": [85, 90, 88]
}

df = pd.DataFrame(
    data,
    index=["S1", "S2", "S3"]
)

print(df)
```

## 8. Columns

Columns represent individual variables or attributes in a DataFrame.

We can access a column using its column name.

```python
data = {
    "Name": ["Alice", "Bob", "Charlie"],
    "Age": [20, 21, 22],
    "Marks": [85, 90, 88]
}

df = pd.DataFrame(data)

print("Columns:")
print(df.columns)

print("\nNames:")
print(df["Name"])
```

```python
print(df[["Name", "Marks"]])
```

## 9. Reading Data

Pandas can read data from different file formats.

Common functions include:

- `pd.read_csv()` for CSV files
- `pd.read_excel()` for Excel files
- `pd.read_json()` for JSON files

```python
data = {
    "Name": ["Alice", "Bob", "Charlie"],
    "Age": [20, 21, 22],
    "Marks": [85, 90, 88]
}

df = pd.DataFrame(data)

df.to_csv("students.csv", index=False)

print("students.csv created successfully")
```

```python
students = pd.read_csv("students.csv")

print(students)
```

## 10. Writing Data

Pandas allows us to save DataFrames to different file formats.

For example, we can use `to_csv()` to save a DataFrame as a CSV file.

```python
data = {
    "Name": ["John", "Sara", "Mike"],
    "Age": [25, 24, 26],
    "Marks": [80, 92, 87]
}

df = pd.DataFrame(data)

df.to_csv("student_data.csv", index=False)

print("Data successfully written to student_data.csv")
```

```python
saved_data = pd.read_csv("student_data.csv")

print(saved_data)
```

## Practical Example

Let's create a student DataFrame and perform some basic operations.

```python
students = pd.DataFrame({
    "Name": ["Alice", "Bob", "Charlie", "David"],
    "Age": [20, 21, 22, 20],
    "Marks": [85, 90, 78, 92]
})

print("Student Data:")
print(students)

print("\nAverage Marks:")
print(students["Marks"].mean())

print("\nStudents with marks greater than 80:")
print(students[students["Marks"] > 80])
```
