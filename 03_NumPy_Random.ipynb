# NumPy Random Module

NumPy provides the `random` module for generating random numbers and performing random sampling.

The NumPy random module is useful in simulations, data analysis, machine learning, probability experiments, and testing.

In this notebook, we will learn about random numbers, random seeds, random integers, random choices, shuffling, permutations, and random probability distributions.

```python
import numpy as np
```

## 1. Random Numbers

The `np.random.rand()` function generates random floating-point numbers between 0 and 1.

Each time the code is executed, different random values may be generated.

```python
random_numbers = np.random.rand(5)

print("Random numbers:", random_numbers)
```

```python
random_matrix = np.random.rand(2, 3)

print("Random 2D array:")
print(random_matrix)
```

## 2. Random Seed

A random seed is used to make random number generation reproducible.

When the same seed is used, NumPy generates the same sequence of random numbers.

This is useful when we want to reproduce the same results during experiments and testing.

```python
np.random.seed(42)

numbers1 = np.random.rand(5)

print("First sequence:")
print(numbers1)
```

```python
np.random.seed(42)

numbers2 = np.random.rand(5)

print("Second sequence:")
print(numbers2)
```

```python
print("Are both sequences equal?", np.array_equal(numbers1, numbers2))
```

## 3. Random Integers

The `np.random.randint()` function generates random integers within a specified range.

The upper limit is excluded.

```python
random_integers = np.random.randint(1, 10, size=5)

print("Random integers:", random_integers)
```

```python
random_matrix = np.random.randint(1, 100, size=(3, 3))

print("Random integer matrix:")
print(random_matrix)
```

## 4. Random Choice

The `np.random.choice()` function randomly selects elements from an array.

It can be used for random sampling.

```python
colors = np.array(["Red", "Green", "Blue", "Yellow"])

choice = np.random.choice(colors)

print("Random choice:", choice)
```

```python
choices = np.random.choice(colors, size=3)

print("Random choices:", choices)
```

```python
choices = np.random.choice(colors, size=5, replace=True)

print("Random choices with replacement:", choices)
```

## 5. Shuffle

The `np.random.shuffle()` function randomly changes the order of elements in an array.

The shuffle operation modifies the original array.

```python
numbers = np.array([1, 2, 3, 4, 5])

print("Before shuffle:")
print(numbers)

np.random.shuffle(numbers)

print("After shuffle:")
print(numbers)
```

## 6. Permutation

The `np.random.permutation()` function returns a randomly rearranged copy of an array.

Unlike `shuffle()`, permutation does not modify the original array.

```python
numbers = np.array([1, 2, 3, 4, 5])

permuted = np.random.permutation(numbers)

print("Original array:")
print(numbers)

print("Permuted array:")
print(permuted)
```

## 7. Random Distributions

A probability distribution describes how random values are distributed.

NumPy provides several probability distributions through the `np.random` module.

Some commonly used distributions are:

- Normal distribution
- Uniform distribution
- Binomial distribution
- Poisson distribution

### Normal Distribution

A normal distribution is a continuous probability distribution that is commonly used in statistics and data analysis.

It is defined using a mean and standard deviation.

```python
normal_data = np.random.normal(loc=50, scale=10, size=1000)

print("First 10 values:")
print(normal_data[:10])
```

```python
print("Mean:", np.mean(normal_data))
print("Standard deviation:", np.std(normal_data))
```

### Uniform Distribution

A uniform distribution generates values where each value within a specified range has an equal probability of being selected.

```python
uniform_data = np.random.uniform(1, 10, size=10)

print("Uniform random values:")
print(uniform_data)
```

### Binomial Distribution

The binomial distribution models the number of successes in a fixed number of independent trials.

It requires:

- Number of trials
- Probability of success

```python
binomial_data = np.random.binomial(n=10, p=0.5, size=10)

print("Binomial values:")
print(binomial_data)
```

### Poisson Distribution

The Poisson distribution is commonly used to model the number of events occurring within a fixed interval of time or space.

```python
poisson_data = np.random.poisson(lam=3, size=10)

print("Poisson values:")
print(poisson_data)
```

## Practical Example

Suppose we want to randomly select students from a class for an activity.

We can use NumPy's random choice function to select students randomly.

```python
students = np.array([
    "Student1",
    "Student2",
    "Student3",
    "Student4",
    "Student5"
])

np.random.seed(42)

selected_students = np.random.choice(students, size=2, replace=False)

print("Selected students:")
print(selected_students)
```
