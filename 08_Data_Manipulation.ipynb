# Data Manipulation with Pandas

Data manipulation involves modifying, transforming, and organizing data to make it suitable for analysis.

In this notebook, we will learn how to rename columns, add and remove rows and columns, transform values, and handle duplicate records using Pandas.

```python
import pandas as pd
import numpy as np
```

```python
data = {
    "Name": ["Alice", "Bob", "Charlie", "David", "Alice"],
    "Age": [20, 21, 22, 23, 20],
    "City": ["Hyderabad", "Chennai", "Hyderabad", "Bangalore", "Hyderabad"],
    "Marks": [85, 90, 78, 92, 85]
}

df = pd.DataFrame(data)

print(df)
```

## 1. Rename Columns

The `rename()` method is used to change the names of rows or columns.

We can rename one or multiple columns without changing the actual data.

```python
renamed_df = df.rename(columns={
    "Name": "Student_Name",
    "Marks": "Score"
})

print(renamed_df)
```

```python
df_all = df.copy()

df_all.columns = ["Student_Name", "Age", "City", "Score",]

print(df_all)
```

## 2. Drop Rows

The `drop()` method can be used to remove rows from a DataFrame.

Rows are usually removed using their index labels.

```python
dropped_row = df.drop(index=0)

print(dropped_row)
```

```python
dropped_rows = df.drop(index=[0, 2])

print(dropped_rows)
```

## 3. Drop Columns

The `drop()` method can also be used to remove columns.

We specify the column name and use `axis=1`.

```python
dropped_column = df.drop("Age", axis=1)

print(dropped_column)
```

```python
dropped_columns = df.drop(
    columns=["Age", "City"]
)

print(dropped_columns)
```

## 4. Insert Columns

The `insert()` method allows us to add a new column at a specific position in a DataFrame.

```python
df_insert = df.copy()

df_insert.insert(
    1,
    "Country",
    "India"
)

print(df_insert)
```

```python
df_insert = df.copy()

df_insert.insert(
    4,
    "Passed",
    df_insert["Marks"] >= 40
)

print(df_insert)
```

## 5. assign()

The `assign()` method is used to create new columns or modify existing columns.

It returns a new DataFrame instead of modifying the original DataFrame directly.

```python
df_assigned = df.assign(
    Passed=df["Marks"] >= 40
)

print(df_assigned)
```

```python
df_assigned = df.assign(
    Bonus_Marks=df["Marks"] + 5
)

print(df_assigned)
```

## 6. apply()

The `apply()` method applies a function to values across a Series or DataFrame.

It is useful for performing custom transformations.

```python
df_apply = df.copy()

df_apply["Marks_Doubled"] = df_apply["Marks"].apply(
    lambda x: x * 2
)

print(df_apply)
```

```python
def calculate_grade(marks):
    if marks >= 90:
        return "A"
    elif marks >= 80:
        return "B"
    else:
        return "C"

df_apply["Grade"] = df_apply["Marks"].apply(calculate_grade)

print(df_apply)
```
